```python
from ultralytics import YOLO
import cv2
import numpy as np
```

```python
# ===== CONFIG =====
INPUT_VIDEO = "input.avi"
OUTPUT_VIDEO = "output_detected.avi"
MODEL_WEIGHTS = "yolov8n.pt"   # poți schimba în yolov8s.pt pentru mai bun
CONF_THRESHOLD = 0.3
```

```python
# COCO class id pentru "sports ball"
BALL_CLASS_ID = 32
```

```python
# ===== LOAD MODEL =====
model = YOLO(MODEL_WEIGHTS)
```

```python
# ===== VIDEO IO =====
cap = cv2.VideoCapture(INPUT_VIDEO)

fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, fps, (width, height))
```

```python
# ===== PROCESS =====
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # inferență
    results = model(frame, verbose=False)[0]

    for box in results.boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])

        if cls == BALL_CLASS_ID and conf > CONF_THRESHOLD:
            x1, y1, x2, y2 = map(int, box.xyxy[0])

            # centru
            cx = (x1 + x2) // 2
            cy = (y1 + y2) // 2

            # aproximăm raza din bounding box
            radius = int(max(x2 - x1, y2 - y1) / 2)

            # desen cerc
            cv2.circle(frame, (cx, cy), radius, (0, 255, 0), 2)

            # punct central
            cv2.circle(frame, (cx, cy), 4, (0, 0, 255), -1)

            # opțional: confidence
            cv2.putText(frame, f"{conf:.2f}", (x1, y1 - 5),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)

    out.write(frame)

    # debug vizual (poți scoate)
    cv2.imshow("Detection", frame)
    if cv2.waitKey(1) == 27:
        break
```

```python
# ===== CLEANUP =====
cap.release()
out.release()
cv2.destroyAllWindows()
```
